```python
class Student:
    def __init__(self, users, details):
        self.users_data = users
        self.details_data = details

    def users(self):
        return self.users_data

    def details(self):
        return self.details_data


s = Student(
    ["dheraj", "kunal", "harsh", "praveen"],
    {
        "address": "kukas",
        "college": "arya",
        "loc": "jaipur"
    }
)

print(s.users())
print(s.details())
```
