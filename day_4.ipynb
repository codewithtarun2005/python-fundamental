```python

def square(x):
  return x*x


#map (2 argument)
l = [10,20,30]
l1 = list(map(square,l))
print(l1)

#exampel
#map (2 argument)
l = [10,20,30]
l1 = list(map(lambda x:x*x,l))
print(l1)

#alternative | without buult in fn
l1 = []
for i in range(len(l)):
  l1.append(l[i] * l[i])
print(l1)

#filter (2 argument)
def hello(x):
  return x.startswith('a')

l = ["apple","banana","cat","dog"]
l1 = list(filter(hello,l))
print(l1)

#exampel
l1 = []
for i in l:
  if 'a' == i[-1]:
    l1.append(i)
    print(l1)

```

```python
#nested list

l = [10,20,30,["hello","hello1","hello2",[True,False]]]
print(l)
print(l[0])
print(l[1])
print(l[2])
print(l[3])
print(l[3][0])
print(l[3][1])
print(l[3][2])
print(l[3][3])
#slicing
l1 = l[1:3]
print(l1)
print(l[3][0:3])
```
